***Calculating information gain for all the variables***
- This is one of the ways to feature selection
- This can be used for XGboost,Gradient boosting and other tree based algorithms
- higher the info_gain better the feature for selection

```python
import math
```

```python
#Calculating Entropy
def calc_entropy(column):
    counts = np.bincount(column)
    probabilities = counts / len(column)
    return -sum(p * math.log(p, 2) for p in probabilities if p > 0)

#Calculating Information Gain
def calc_information_gain(data, feature, target):
    original_entropy = calc_entropy(data[target])
    values = data[feature].unique()

    weighted_entropy = 0
    for value in values:
        subset = data[data[feature] == value]
        weight = len(subset) / len(data)
        weighted_entropy += weight * calc_entropy(subset[target])
    
    return original_entropy - weighted_entropy

#Calculating info gain for all features
def get_information_gain_df(df, target):
    info_gain = {
        col: calc_information_gain(df, col, target)
        for col in df.columns if col != target
    }
    result_df = pd.DataFrame(info_gain.items(), columns=['Feature', 'Info Gain'])
    return result_df[result_df['Info Gain'] >= 0]

#Calling the function
df_inf_gain = get_information_gain_df(df_train, 'TARGET')
print(df_inf_gain)
```

# Using Xgboost

```python
#sanitize the column names 
X_train.columns = X_train.columns.astype(str).str.replace(r"[<>\[\]]", "_", regex=True)
X_test.columns = X_test.columns.astype(str).str.replace(r"[<>\[\]]", "_", regex=True)
X_val.columns = X_val.columns.astype(str).str.replace(r"[<>\[\]]", "_", regex=True)
```

```python
X_train.columns
```

```python
# Xgboost model
```

```python
# Count of classes in training data
num_0 = sum(y_train == 0)
num_1 = sum(y_train == 1)

# Ratio of majority to minority class
scale_pos_weight = num_0 / num_1
print("scale_pos_weight:", scale_pos_weight)
```

```python
y_train
```

```python
X_train
```

```python
# Flatten column names in case they're tuples (multi-index from one-hot)
def flatten_columns(df):
    df.columns = ['_'.join(map(str, col)) if isinstance(col, tuple) else str(col) for col in df.columns]
    return df

X_train = flatten_columns(X_train)
X_test = flatten_columns(X_test)
X_val = flatten_columns(X_val)
```

```python
#Converting it into numeric values
X_train = X_train.astype(float)
X_test = X_test.astype(float)
X_val = X_val.astype(float)
```

```python
import xgboost as xgb
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import pandas as pd

# Initialize the classifier with scale_pos_weight
xgb_cl = xgb.XGBClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=5,
    subsample=0.8,
    colsample_bytree=0.8,
    use_label_encoder=False,
    eval_metric='logloss',
    scale_pos_weight=scale_pos_weight,  # <--- This handles imbalance
    random_state=42
)
```

```python
# Train the model on training data
xgb_cl.fit(X_train.values, y_train.values)

# Predict on train, test, and validation data
y_train_pred = xgb_cl.predict(X_train.values)
y_test_pred = xgb_cl.predict(X_test.values)
y_val_pred = xgb_cl.predict(X_val.values)

# Evaluate accuracy
print("Train Accuracy:", accuracy_score(y_train, y_train_pred))
print("Test Accuracy:", accuracy_score(y_test, y_test_pred))
print("Validation Accuracy:", accuracy_score(y_val, y_val_pred))

# Detailed report
print("\nClassification Report (Validation):")
print(classification_report(y_val, y_val_pred))

print("\nConfusion Matrix (Validation):")
print(confusion_matrix(y_val, y_val_pred))
```

```python
# Feature importance
feat_imp = pd.DataFrame({
    'Feature': X_train.columns,
    'Importance': xgb_cl.feature_importances_ * 100
}).sort_values(by='Importance', ascending=False).reset_index(drop=True)

print("\nTop Feature Importances:")
print(feat_imp.head(10))
```

```python
# Define a threshold
threshold = 1.0  # Features with importance < 1% will be removed

# Filter the features to keep
important_features = feat_imp[feat_imp['Importance'] >= threshold]['Feature'].tolist()

# Reduce X_train, X_test, and X_val to only the important features
X_train_reduced = X_train[important_features]
X_test_reduced = X_test[important_features]
X_val_reduced = X_val[important_features]
```

```python
# Train again using only important features
xgb_cl.fit(X_train_reduced.values, y_train)

# Predict again
y_train_pred = xgb_cl.predict(X_train_reduced.values)
y_test_pred = xgb_cl.predict(X_test_reduced.values)
y_val_pred = xgb_cl.predict(X_val_reduced.values)

# Evaluate again
print("Training Accuracy (Reduced):", accuracy_score(y_train, y_train_pred))
print("Testing Accuracy (Reduced):", accuracy_score(y_test, y_test_pred))
print("Validation Accuracy (Reduced):", accuracy_score(y_val, y_val_pred))
```
